```pyspark
# Initialize Hail against the running SparkContext from Livy/EMR Notebooks
import hail as hl
hl.init(sc, log="/tmp/hail.log")
```

```pyspark
# source
vds_prefix = "s3://precise-scratch/goypav/SG10K_Health/VDS/"
vds_step4_prefix = vds_prefix + "step4/"

# input
manifest_uri = "s3://precise-scratch/goypav/SG10K_Health/VDS/vds_step3_manifest_SG10K_Health_batch1_5_10321.txt"

# output
release_name = "SG10K_Health_r1.0.0.n10321"

out_uri = vds_step4_prefix + release_name + ".vds"
tmp_uri = vds_step4_prefix + "tmp/"
log_uri = vds_step4_prefix + release_name + ".log"
input_manifest_uri = vds_step4_prefix + release_name + ".inputs.txt"
```

```pyspark
# -----------------------------
# 1. Read manifest of VDS paths
# -----------------------------
# # create manifest vds_step3_manifest_SG10K_Health_batch1_5_10321
# for batch in batch1_1000 batch2_999 batch3_999 batch4_5_7323; do
#     aws s3 ls s3://precise-scratch/goypav/SG10K_Health/VDS/step3/${batch}/ \
#     | awk -v p="s3://precise-scratch/goypav/SG10K_Health/VDS/step3/${batch}/" \
#         '$NF ~ /\.vds\/?$/ { print p $NF }'
# done \
# | sort \
# > vds_step3_manifest_SG10K_Health_batch1_5_10321.txt

# ### copy to s3
# aws s3 cp vds_step3_manifest_SG10K_Health_batch1_5_10321.txt s3://precise-scratch/goypav/SG10K_Health/VDS/ --dryrun
    
fs = hl.current_backend().fs

with fs.open(manifest_uri) as f:
    vds_paths = [
        line.strip()
        for line in f
        if line.strip()
    ]

vds_paths = sorted(vds_paths)

print(f"Found {len(vds_paths):,} VDSes")
print("\nFirst 5 VDSes:")
print(*vds_paths[:5], sep="\n")
```

```pyspark
# ============================================================
# COMBINE
# ============================================================

import time
import traceback
import gc
from datetime import datetime

branch_factor = 6
target_records = 200_000

def exists(path):
    try:
        fs.ls(path)
        return True
    except Exception:
        return False

start = time.time()
log_lines = []

def log(msg):
    line = f"[{datetime.now():%Y-%m-%d %H:%M:%S}] {msg}"
    print(line)
    log_lines.append(line)

try:
    log("Starting STEP4 final combine")
    log(f"Number of input VDSes: {len(vds_paths)}")
    log(f"Output URI: {out_uri}")
    log(f"Temp URI: {tmp_uri}")
    log(f"branch_factor: {branch_factor}")
    log(f"target_records: {target_records}")

    with fs.open(input_manifest_uri, "w") as f:
        f.write("\n".join(vds_paths) + "\n")

    if exists(out_uri):
        log("Output already exists, skipping combine")

    else:
        combiner = hl.vds.new_combiner(
            output_path=out_uri,
            temp_path=tmp_uri,
            vds_paths=vds_paths,
            use_genome_default_intervals=True,
            reference_genome="GRCh38",
            branch_factor=branch_factor,
            target_records=target_records,
            gvcf_save_filters=True,
        )

        combiner.run()

        elapsed = time.time() - start
        log("Completed STEP4 final combine")
        log(f"Runtime: {elapsed:.1f} seconds ({elapsed / 60:.2f} minutes)")

except Exception as e:
    elapsed = time.time() - start
    log("FAILED STEP4 final combine")
    log(f"Runtime before failure: {elapsed:.1f} seconds ({elapsed / 60:.2f} minutes)")
    log(f"Error: {repr(e)}")
    log(traceback.format_exc())
    raise

finally:
    with fs.open(log_uri, "w") as f:
        f.write("\n".join(log_lines) + "\n")

    if "combiner" in locals():
        del combiner

    gc.collect()
```

```pyspark
# ============================================================
# OUTPUT CHECK
# ============================================================

print(f"Reading final VDS: {out_uri}")

vds = hl.vds.read_vds(out_uri)

observed_samples = vds.n_samples()

expected_samples = 0
for p in vds_paths:
    expected_samples += hl.vds.read_vds(p).n_samples()

n_var_rows = vds.variant_data.count_rows()
n_ref_rows = vds.reference_data.count_rows()

status = "OK" if observed_samples == expected_samples else "MISMATCH"

print(
    f"{status}: "
    f"expected_samples={expected_samples:,}, "
    f"observed_samples={observed_samples:,}"
)

print(f"Variant rows:   {n_var_rows:,}")
print(f"Reference rows: {n_ref_rows:,}")
print(f"Variant partitions:   {vds.variant_data.n_partitions():,}")
print(f"Reference partitions: {vds.reference_data.n_partitions():,}")
```
