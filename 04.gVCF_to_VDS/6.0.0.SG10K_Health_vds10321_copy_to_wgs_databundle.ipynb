```pyspark
# Initialize Hail against the running SparkContext from Livy/EMR Notebooks
import hail as hl
hl.init(sc, log="/tmp/hail.log")
```

```pyspark
# source
vds_prefix = "s3://precise-scratch/goypav/SG10K_Health/VDS/"
vds_step4_prefix = vds_prefix + "step4/"
r600_wgs_uri = 's3://precise-wgs-databundle/byob/sg10k-releases/SG10K_Health_r6/'

# # input
vds_uri = vds_step4_prefix + "SG10K_Health_r1.0.0.n10321.vds"

# ------------------- outputs --------------------------------------------------
# intermediate checkpoints

# # output
vds_r600_wgs_uri = r600_wgs_uri + "SG10K_Health_r6.0.0.n10321.vds"

overwrite = False
```

## check VDS

```pyspark
# read VDS
vds = hl.vds.read_vds(vds_uri)
```

```pyspark
# check reference_data
vds.reference_data.describe()
```

```pyspark
# check variant_data
vds.variant_data.describe()
```

```pyspark
# Count rows/cols in the variant_data MT
print(f"Reference genome: {vds.variant_data.locus.dtype.reference_genome.name}")
print(f"Number of samples: {vds.n_samples():,}")
print(f"Total number of variants: {vds.variant_data.count_rows():,}")
print(f"Number of variant partitions: {vds.variant_data.n_partitions():,}")
```

```pyspark
hl.eval(vds.reference_data.ref_block_max_length)
```

## copy VDS to wgs

```pyspark
####################################
# copy VDS
####################################
vds.write(vds_r600_wgs_uri, overwrite=False)
print("Wrote VDS to:", vds_r600_wgs_uri)

# 1h - 1000 CPUs
```
