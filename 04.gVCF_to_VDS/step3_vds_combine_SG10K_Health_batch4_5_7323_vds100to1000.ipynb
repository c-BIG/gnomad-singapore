```pyspark
# source
vds_prefix = "s3://precise-scratch/goypav/SG10K_Health/VDS/"
vds_step3_prefix = vds_prefix + "step3/batch4_5_7323/"

# input
manifest_uri = "s3://precise-scratch/goypav/SG10K_Health/VDS/vds_step2_manifest_SG10K_Health_batch4_5_7323.txt"

# output

```

```pyspark
# -----------------------------
# 1. Read manifest of VDS paths
# -----------------------------
fs = hl.current_backend().fs

with fs.open(manifest_uri) as f:
    vds_paths = [
        line.strip()
        for line in f
        if line.strip()
    ]

vds_paths = sorted(vds_paths)

print(f"Found {len(vds_paths):,} VDSes")
print("\nFirst 5 VDSes:")
print(*vds_paths[:5], sep="\n")
```

```pyspark
import time
import traceback
import gc
from datetime import datetime

# -----------------------------
# 3. Split VDSes into groups of 10
# -----------------------------

step3_batch_size = 10

vds_batches = [
    vds_paths[i:i + step3_batch_size]
    for i in range(0, len(vds_paths), step3_batch_size)
]

print(f"\nNumber of step3 batches: {len(vds_batches)}")

for i, batch in enumerate(vds_batches):
    print(f"step3_batch_{i:04d}: {len(batch)} input VDSes")
```

```pyspark
# ============================================================
# STEP3 SINGLE-BATCH CELL
# Change only this number: 0, 1, 2, 3, 4, 5, 6, or 7
# ============================================================

batch_idx_to_run = 0

branch_factor = 10
target_records = 500_000

logs_prefix = vds_step3_prefix + "logs/"

def exists(path):
    try:
        fs.ls(path)
        return True
    except Exception:
        return False

batch_idx = batch_idx_to_run
batch_paths = vds_batches[batch_idx]

batch_name = f"step3_batch_{batch_idx:04d}"

out_uri = f"{vds_step3_prefix}{batch_name}.vds"
tmp_uri = f"{vds_step3_prefix}tmp/{batch_name}/"

log_uri = f"{logs_prefix}{batch_name}.log"
input_manifest_uri = f"{logs_prefix}{batch_name}.inputs.txt"

start = time.time()
log_lines = []

def log(msg):
    line = f"[{datetime.now():%Y-%m-%d %H:%M:%S}] {msg}"
    print(line)
    log_lines.append(line)

try:
    log(f"Starting {batch_name}")
    log(f"Number of input VDSes: {len(batch_paths)}")
    log(f"Output URI: {out_uri}")
    log(f"Temp URI: {tmp_uri}")

    with fs.open(input_manifest_uri, "w") as f:
        f.write("\n".join(batch_paths) + "\n")

    if exists(out_uri):
        log("Output already exists, skipping")

    else:
        combiner = hl.vds.new_combiner(
            output_path=out_uri,
            temp_path=tmp_uri,
            vds_paths=batch_paths,
            use_genome_default_intervals=True,
            reference_genome="GRCh38",
            branch_factor=branch_factor,
            target_records=target_records,
            gvcf_save_filters=True,
        )

        combiner.run()

        elapsed = time.time() - start
        log(f"Completed {batch_name}")
        log(f"Runtime: {elapsed:.1f} seconds ({elapsed / 60:.2f} minutes)")

except Exception as e:
    elapsed = time.time() - start
    log(f"FAILED {batch_name}")
    log(f"Runtime before failure: {elapsed:.1f} seconds ({elapsed / 60:.2f} minutes)")
    log(f"Error: {repr(e)}")
    log(traceback.format_exc())
    raise

finally:
    with fs.open(log_uri, "w") as f:
        f.write("\n".join(log_lines) + "\n")

    if "combiner" in locals():
        del combiner

    gc.collect()
```
